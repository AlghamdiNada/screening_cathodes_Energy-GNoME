```python
from ase.io import read, write
from ase.visualize import view
from ase.build import make_supercell
```

```python
atoms = read("NaCoO2.cif")
```

```python
view(atoms, viewer='nglview')
```

```python
atoms.get_chemical_formula()
```

```python
supercell = make_supercell(atoms, [[2, 0, 0], [0, 1, 0], [0, 0, 1]])
```

```python
supercell.get_chemical_formula()
```

```python
view(supercell, viewer='nglview')
```

```python
write('2x1x1-NaCoO2.cif', supercell)
```
